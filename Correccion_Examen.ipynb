9.-Producto más vendido por sucursal. ¿Qué producto acumuló más unidades vendidas en cada sucursal? Indique su código, nombre y cantidad vendida. Si hay un empate, incluya todos los productos empatados.

```python
columnas=agrupar_ambos_csv["sucursal"]
filas=[agrupar_ambos_csv["id_producto"],agrupar_ambos_csv["producto"]]
values=agrupar_ambos_csv["cantidad"]
producto_mas_vendido_por_sucursal=pd.crosstab(filas,columnas,values=values,aggfunc="sum").fillna(0)
producto_mas_vendido_por_sucursal
```

```python
for sucursal in producto_mas_vendido_por_sucursal.columns:
    col_sucursal = producto_mas_vendido_por_sucursal[sucursal]
    max_unidades = col_sucursal.max()
    ganadores = col_sucursal[col_sucursal == max_unidades]

    print("--- Sucursal", sucursal, "---")

    for (codigo, nombre), cantidad in ganadores.items():
        print(" ", codigo, "-", nombre, ":", cantidad, "unidades")
```
